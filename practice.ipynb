```python
tokyo_point = (
    35.6895,
    139.6917,
)
```

```python
print(tokyo_point)
```

```python
latitude = tokyo_point[0]
longitude = tokyo_point[1]
```

```python
print(f"Tokyo's latitude is {latitude}")
print(f"Tokyo's longitude is {longitude}")
```

```python
lat, lon = tokyo_point
```

The unpacking of the tuple means the lat gets the variable at the first position in the tuple tokyo_point[0] and the second gets assigned longitude = tokyo_point[1] automagically. The number of variables on the left, must match the number of values inside the tuple.

```python
print(f"Tokyo is located at {lat}N, {lon}E")
```
